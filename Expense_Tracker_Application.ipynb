```python
import pandas as pd

data = {
    "Date":["2026--09-01","2026-09-02","2026-09-03","2026-09-04","2026-09-05"],
    "Category":["Food","Travel","Shopping","Food","Bills"],
    "Amount":[150,80,500,200,1000]
}

df =pd.DataFrame(data)

df
```

```python
total_expense =df["Amount"].sum()
print("Total Expense:",total_expense)
```

```python
category_expense =df.groupby("Category")["Amount"].sum()
print(category_expense)
```

```python
highest_expense =df.loc[df["Amount"].idxmax()]
print("Highest Expense:")
print(highest_expense)
```

```python
lowest_expense = df.loc[df["Amount"].idxmin()]
print("Lowest Expense:")
print(lowest_expense)
```

```python
import matplotlib.pyplot as plt

category_expense.plot(kind="bar")

plt.title("Expense by Category")
plt.xlabel("Category")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.show()
```

```python
average_expense = df["Amount"].mean()

print("Expense Summary")
print("----------------")
print("Total Expense:",total_expense)
print("Average Expense:",average_expense)
print("Highest Expense:",df["Amount"].max())
print("Lowest Expense:",df["Amount"].min())
```

```python
df.to_csv("expense_data.csv",index=False)

print("Expense data saved successfully!")
```

```python
import pandas as pd

data={
    "Date":["2026-09-01","2026-09-02","2026-09-03","2026-09-04","2026-09-05"],
    "Category":["Food","Travel","Shopping","Food","Bills"],
    "Amount":[150,80,500,200,1000]
}

df=pd.DataFrame(data)

df["Date"]=pd.to_datetime(df["Date"])
df["Month"]=df["Date"].dt.month

monthly_expense=df.groupby("Month")["Amount"].sum()

print("Monthly Expense:")
print(monthly_expense)
```

```python
monthly_expense.plot(kind="bar")

plt.title("Monthly Expense")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.show()
```

```python
print("Expense Statistics")
print("----------------------")
print(df["Amount"].describe())
```

```python
category_expense.plot(kind="pie",autopct="%1.1f%%")

plt.title("Expense Distribution by Category")
plt.ylabel("")
plt.show()
```

```python
def add_expense(date,category,amount):
  global df

  new_expense =pd.DataFrame({
      "Date":[pd.to_datetime(date)],
      "Category":[[category]],
      "Amount":[amount]
  })

  df =pd.concat([df,new_expense],ignore_index=True)

  print("Expense added successfully!")
```

```python
add_expense("2026-09-06","Food",120)
```

```python
import pandas as pd

data ={
    "Date":[
        "2026-09-01",
        "2026-09-02",
        "2026-09-03",
        "2026-09-04",
        "2026-09-05",
        "2026-09-06",
    ],
    "Category":[
        "Food",
        "Travel",
        "Shopping",
        "Food",
        "Bills",
        "Food"
    ],
    "Amount":[
        150,
        80,
        500,
        200,
        1000,
        120
    ]
}

df=pd.DataFrame(data)

print(df[df["Category"]=="Food"])
```

```python
food_total =df[df["Category"] =="Food"]["Amount"].sum()

print("Total Food Expense:",food_total)
```

```python
category_total =df.groupby("Category")["Amount"].sum()

print("Category-wise Total Expense:")
print(category_total)
```

```python
budget =2500

total_expense =df["Amount"].sum()

print("Montly Budget:",budget)
print("Total Expense:",total_expense)

if total_expense<=budget:
  print("Budget Status:Within Budget")
else:
  print("Budget Status:Over Budget")
```
